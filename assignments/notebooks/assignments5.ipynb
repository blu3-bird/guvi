### Q28. The Monty Hall Problem
  

    Write a Python simulation that plays the Monty Hall game 10,000 times.
    Calculate the win percentage for:  


    Always Switching  

    Never Switching  

    Explain why the two probabilities differ.  


```python
monty_hall_n = 10000
switch = 0
never_switch = 0

doors = [0,1,2]

for i in range(monty_hall_n):

    car_door = np.random.choice(doors)

    player_choice = np.random.choice(doors)

    monty_choice = [
        d for d in doors if d != car_door and d !=player_choice
    ]

    monty_choice = np.random.choice(monty_choice)

    if player_choice == car_door:
        never_switch +=1
    else:
        switch +=1


prob_win = never_switch/monty_hall_n
prob_switch_win = switch / monty_hall_n
prob_switch_win
```
